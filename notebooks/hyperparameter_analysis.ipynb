```python
datasets = sorted([dsfile.stem for dsfile in DATASET_DIR.iterdir()], reverse=True)
# datasets = ['concentric3']

nnbrs_list = [5, 15, 50, 100, 200]
min_dist_list = [0.0, 0.1, 0.3, 0.5, 0.9]

umap_results = {}

for dataset in datasets:
    print(f"Computing UMAP for {dataset}...")
    dsfile = f"{dataset}.npy"

    X = np.load(DATASET_DIR / dsfile)
    labels = np.load(LABELS_DIR / dsfile)

    umap_results[dataset] = {}

    for nn in tqdm(nnbrs_list, position=0, desc='n_neighbors'):
        umap_results[dataset][nn] = {}

        for md in tqdm(min_dist_list, position=1, desc='min_dist'):
            um = UMAP(init='spectral', n_neighbors=nn, min_dist=md)
            Y = um.fit_transform(X)
            score = CADI(X, Y, labels, n_triplets=X.shape[0] * 40)

            umap_results[dataset][nn][md] = {
                "embedding": Y,
                "score": score,
            }

print("Finished computing all UMAP embeddings.")
```

```python
plot_args = {
    'olivetti': {'s': 25, 'cmap': 'tab20'},
    'penguins': {'s': 30},
    'coil20': {'s': 10},
    'matryoshka': {'s': 4, 'cmap': 'tab10', 'alpha':0.5},
    'fashionMNIST': {'s': 1},
    'pendigits': {'s': 10, 'cmap': 'tab10'},
    'usps': {'s': 4, 'cmap': 'tab10'},
    'concentric4': {'s': 8},
    'concentric3': {'s': 8}
}
```

```python
import matplotlib as mpl

# Load the colormap for the borders
score_cmap = mpl.colormaps["vanimo_r"]

# Global typography for quality output
plt.rcParams.update({
    "font.size": 11,
    "axes.titlesize": 12,
    "axes.labelsize": 12,
    "figure.titlesize": 14,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
})

dataset = 'concentric3'
dsfile = f"{dataset}.npy"
labels = np.load(LABELS_DIR / dsfile)

nnbrs_list = sorted(umap_results[dataset].keys())
min_dist_list = sorted(next(iter(umap_results[dataset].values())).keys())

n_cols = len(nnbrs_list)
n_rows = len(min_dist_list)

# Publication-friendly sizing
fig_w = 3.2 * n_cols
fig_h = 3.2 * n_rows

fig, axes = plt.subplots(
    n_rows, n_cols,
    figsize=(fig_w, fig_h),
    constrained_layout=False
)

# Ensure axes always 2D
if n_rows == 1:
    axes = axes[None, :]
if n_cols == 1:
    axes = axes[:, None]

# Collect all scores to normalize border colors
all_scores = np.array([
    umap_results[dataset][nn][md]["score"]
    for nn in nnbrs_list for md in min_dist_list
])
smin, smax = all_scores.min(), all_scores.max()


for i, md in enumerate(min_dist_list):
    for j, nn in enumerate(nnbrs_list):

        ax = axes[i, j]
        entry = umap_results[dataset][nn][md]

        Y = entry["embedding"]
        score = entry["score"]

        # Plot embeddings
        ax.scatter(*Y.T, c=labels,
                    **plot_args.get(dataset,
                                    {'s': 4, 'cmap': 'rainbow'})
                    )
        
        # Determine square limits so aspect ratio is preserved
        min_x, max_x = Y[:,0].min(), Y[:,0].max()
        min_y, max_y = Y[:,1].min(), Y[:,1].max()
        padding = 0.05  # 5% padding
        span = max(max_x - min_x, max_y - min_y)
        span *= (1 + 2 * padding)
        mid_x = 0.5 * (min_x + max_x)
        mid_y = 0.5 * (min_y + max_y)

        ax.set_xlim(mid_x - span/2, mid_x + span/2)
        ax.set_ylim(mid_y - span/2, mid_y + span/2)
        ax.set_aspect("equal", adjustable="box")

        # Remove ticks
        ax.set_xticks([])
        ax.set_yticks([])

        # Border color
        normalized_score = (score - smin) / (smax - smin)
        # Normalize between 0.2 and 0.8 to prevent too light colors
        normalized_score = 0.2 + 0.6 * normalized_score
        border_color = score_cmap(normalized_score)

        for spine in ax.spines.values():
            spine.set_visible(True)
            spine.set_linewidth(2.0)
            spine.set_color(border_color)

        # Column labels (top row)
        if i == 0:
            ax.set_title(f"n_neighbors = {nn}", pad=12, fontsize=20)

        # Row labels (left column)
        if j == 0:
            ax.set_ylabel(f"min_dist = {md}", labelpad=12, fontsize=20)

        # Score under each subplot, bold + color-matched
        ax.text(0.5, -0.01,
                f"{score:.3f}",
                ha="center", va="top",
                transform=ax.transAxes,
                fontsize=20, fontweight="bold",
                color=border_color)

fig.suptitle(f"{dataset.upper()}", y=1.01)


fig.tight_layout()
plt.show()
```

```python
for dataset in datasets:
    nnbrs_list = sorted(umap_results[dataset].keys())
    min_dist_list = sorted(next(iter(umap_results[dataset].values())).keys())

    # Build score matrix
    score_matrix = np.zeros((len(min_dist_list), len(nnbrs_list)))

    for i, md in enumerate(min_dist_list):
        for j, nn in enumerate(nnbrs_list):
            score_matrix[i, j] = umap_results[dataset][nn][md]["score"]

    fig, ax = plt.subplots(figsize=(1.8 * len(nnbrs_list), 1.5 * len(min_dist_list)))

    im = ax.imshow(score_matrix, aspect='auto', cmap='Oranges')

    # Tick labels
    ax.set_xticks(np.arange(len(nnbrs_list)))
    ax.set_xticklabels(nnbrs_list)
    ax.set_xlabel("n_neighbors")

    ax.set_yticks(np.arange(len(min_dist_list)))
    ax.set_yticklabels(min_dist_list)
    ax.set_ylabel("min_dist")

    # Add score text
    for i in range(len(min_dist_list)):
        for j in range(len(nnbrs_list)):
            ax.text(j, i, f"{score_matrix[i, j]:.4f}",
                    ha="center", va="center",
                    color="white" if score_matrix[i, j] > score_matrix.mean() else "black")

    cbar = fig.colorbar(im, ax=ax)
    cbar.set_label('CADI', rotation=270, loc='center', labelpad=20)

    fig.suptitle(f"CADI Score Heatmap - {dataset}", y=1.02)

    plt.tight_layout()
    # fig.savefig(heatmaps_dir / f"{dataset}.pdf", format='pdf')
    plt.show()
```
